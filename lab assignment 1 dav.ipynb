```python
import pandas as pd
df=pd.read_csv("earthquake_1995-2023.csv")
print("rows and columns",df.shape)
df.head()
```

```python
df.info()
print(df.duplicated().sum())
dupe_mask=df.duplicated(subset=["title","date_time"],keep=False)
print("rows involved in a title+data_time",dupe_mask.sum())
df[dupe_mask].sort_values("title")[["title","date_time","magnitude","country"]]
before=len(df)
df=df.drop_duplicates(subset=["title","date_time"],keep="first").reset_index(drop=True)
after=len(df)
print(f"record{before-after} duplicate row(s)")
print("rows now :",after)
import numpy as np
for col in ["cdi","gap","dmin","nst"]:
  df[col]=df[col].replace(0,np.nan)
print("missing values now visible")
print(df[["cdi","gap","dmin","nst"]].isnull().sum())
```

```python
missing_count = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percent
})

print(missing_report)
```

```python
import matplotlib.pyplot as plt

for col in ["cdi", "gap", "nst", "dmin"]:
    print(col, "skewness:", df[col].skew())

    plt.figure(figsize=(6, 4))
    plt.hist(df[col].dropna(), bins=30)
    plt.title("Distribution of " + col)
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
df["alert"].value_counts().plot(kind="bar")
plt.xlabel("Alert Level")
plt.ylabel("Number of Earthquakes")
plt.title("Earthquakes by Alert Level")
plt.show()
```

```python
for col in ["cdi", "gap", "nst", "dmin"]:
    print("\n", col)
    print("Mean  :", df[col].mean())
    print("Median:", df[col].median())
    print("Skewness:", df[col].skew())
print("Missing alert:", df["alert"].isnull().sum())

alert_mode = df["alert"].mode()[0]

print("Alert mode:", alert_mode)

df["alert"] = df["alert"].fillna(alert_mode)

print("Missing alert after filling:",
      df["alert"].isnull().sum())
```

```python
for col in ["cdi", "gap", "nst", "dmin"]:
    df[col] = df[col].fillna(df[col].median())

print("Numeric missing values filled")
categorical_cols = ["alert", "location", "continent", "country"]

for col in categorical_cols:
    df[col] = df[col].fillna("Unknown")

print("Categorical missing values filled")
print("Remaining missing values:")
print(df.isnull().sum())

print("\nTotal missing values:", df.isnull().sum().sum())
```
